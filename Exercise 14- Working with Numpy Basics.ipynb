##### Introduction

This project is focused on the Temperature records of a weather station.

Given a weather station has three locations:
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

Task is to create NumPy arrays for all three locations, and answer the business questions thay follows


```python
# the code below is to demonstrate the import of NumPy into our notebook workspace for this purpose of the
# MINI PROJECT CHALLENGE

import numpy as np

# Note: if numpy has been succesfully imported, after running this cell/code, no error message would appear.
#       but if Numpy import wasn't successful, there would be an error message below this cell.
```

```python
# the code below is to create weather records for three stations
# first we create the python list for each record

# python list
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

# then we convert the python lists to numpy arrays
Lagos = np.array(lagos)
Abuja = np.array(abuja)
Ibadan = np.array(ibadan)

print (Lagos)
print (Abuja)
print (Ibadan)
# the subsequent codes below this cell are to answer business questions from the record of temperatures above.
```

#### Challenge 1: Which Location recorded the highest temperature on each day ?

```python
# since there are five records for each station, therefore, each record represents a day.
# below are the numPy arrays of the recorded temperatures for each station.
Lagos
Abuja
Ibadan

# the code below is to find the highest temperature for each day.

for day in range (5):
    highest_temp = max(Lagos[day],Abuja[day],Ibadan[day])

    if highest_temp == Lagos[day]:
        location = "Lagos"

    elif highest_temp == Abuja[day]:
        location = "Abuja"

    else:
        location = "Ibadan"


    print ("Highest Temperature for Day",day + 1,":", location,"->", highest_temp)
```

#### Challenge 2: Increase every Temperature record by 1

```python
# the code below is to increase every temperature record for each station by 1 degree.
# to do this, we do not need to use a loop, as numpy array as an inbuild element-wise operation ability.
# therefore;

Lagos_plus1 = Lagos + 1
Abuja_plus1 = Abuja + 1
Ibadan_plus1 = Ibadan + 1

# the code below is to display the result of the increased temperature records for each station location
print (Lagos_plus1)
print (Abuja_plus1)
print (Ibadan_plus1)
```

#### Challenge 3: Compare the Temperatures of Lagos and Abuja

```python
# the code below is to compare the temperatures of Lagos and Abuja Station
for day in range (5):
    if Lagos[day] > Abuja[day]:
        print ("Day", day+1, ": Lagos was Hotter")

    elif Abuja[day] > Lagos[day]:
        print ("Day", day+1,": Abuja was Hotter")

    else:
        print ("Day", day+1, ": the Temperature of both locations were the same")
```
